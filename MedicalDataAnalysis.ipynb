```python
import pandas as pd
```

```python
#loading the dataset
df = pd.read_csv('Medicaldataset.csv')
```

```python
#displaying the first 5 rows
df.head()
```

```python
#checking the dataset structure
df.info() #show column names, datatypes and non-null coounts
df.describe()  # Basic statistical summary
df.isnull().sum()  # Check for missing values
```

```python
#there is no null values in my dataset but to drop the null values
df.dropna(inplace=True)  # Remove rows with missing values
```

```python
#computing basic statistics
df.describe()
```

```python
#grouping and computing the mean
df.groupby("Gender")["Heart rate"].mean()
```

```python
#Identify interesting patterns For example, comparing blood pressure values based on heart attack status:
df.groupby("Result")[["Systolic blood pressure", "Diastolic blood pressure"]].mean()
```

```python
#Data Visualization
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Line Chart: Heart Rate vs Age
plt.plot(df["Age"], df["Heart rate"], marker="o")
plt.xlabel("Age")
plt.ylabel("Heart Rate")
plt.title("Heart Rate Variation by Age")
plt.show()
```

```python
#Bar Chart: Blood Pressure Comparison Across Genders
sns.barplot(x="Gender", y="Systolic blood pressure", data=df)
plt.title("Average Systolic Blood Pressure by Gender")
plt.show()
```

```python
#Histogram: Distribution of Heart Rate
plt.hist(df["Heart rate"], bins=15, edgecolor="black")
plt.xlabel("Heart Rate")
plt.ylabel("Frequency")
plt.title("Distribution of Heart Rate")
plt.show()
```

```python
#Scatter Plot: Relationship Between Systolic & Diastolic Blood Pressure
plt.scatter(df["Systolic blood pressure"], df["Diastolic blood pressure"])
plt.xlabel("Systolic Blood Pressure")
plt.ylabel("Diastolic Blood Pressure")
plt.title("Blood Pressure Relationship")
plt.show()
```
